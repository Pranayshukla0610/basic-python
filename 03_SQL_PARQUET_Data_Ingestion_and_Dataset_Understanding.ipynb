```python
all_records=[]

for page in range(1,6):

  params = {
      "page":page,
      "limit":100
  }

  response = requests.get(
      url,
      params=params,
      timeout=10
  )

  response.raise_for_status()

  data = response.json()

  all_records.extend(data)
```

#PARQUET

```python
import pandas as pd

df = pd.read_json('/content/ecommerce_sales_34500-selected-columns.csv')

df.head()
```

```python
df.to_parquet(
    "/content/ecommerce_sales_34500-selected-columns.parquet",
    index=False
)
```

```python
parquet_df = pd.read_parquet(
    "/content/ecommerce_sales_34500-selected-columns.parquet",
    columns=[
        "product_id",
        "quantity"
    ]
)

parquet_df.head()
```

```python
csv_size = os.path.getsize(
    "/content/ecommerce_sales_34500-selected-columns.csv"
)

parquet_size = os.path.getsize(
    "/content/ecommerce_sales_34500-selected-columns.parquet"
)

print(
    "CSV:",
    csv_size / (1024 * 1024),
    "MB"
)

print(
    "Parquet:",
    parquet_size / (1024 * 1024),
    "MB"
)
```
