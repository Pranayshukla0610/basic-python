```python
#Find the Top 2 Performing Employees in Each Department Based on Performance Score
import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Department": ["HR", "HR", "IT", "IT", "IT", "Sales", "Sales", "Sales"],
    "Performance": [85, 92, 88, 95, 91, 80, 96, 89]
})
```

```python
df['Rank'] = df.groupby('Department')['Performance'].rank(method='dense',ascending=False)

result = df[df['Rank'] <=2].sort_values(by=['Department','Rank'])

print(result)
```

```python
#Find Employees Whose Salary is Greater Than the Department Median

import pandas as pd

df = pd.DataFrame({
    "Employee":["A","B","C","D","E","F","G","H"],
    "Department":["HR","HR","HR","IT","IT","IT","Sales","Sales"],
    "Salary":[40000,50000,80000,70000,90000,85000,45000,60000]
})
```

```python
df["med_dept"] = df.groupby('Department')['Salary'].transform('median')

result = df[df['Salary'] > df['med_dept']]
print(result)
```

```python
#Find Employees Whose Salary is More Than Their Manager's Salary

import pandas as pd

df = pd.DataFrame({
    "EmpID":[1,2,3,4,5],
    "Employee":["Alice","Bob","Charlie","David","Eva"],
    "ManagerID":[None,1,1,2,2],
    "Salary":[90000,60000,95000,70000,65000]
})
```

```python
manager_salary = df[['EmpID','Salary']].rename(
    columns={
        "EmpID":"ManagerID",
        "Salary":"ManagerSalary"
    }
)

result = df.merge(manager_salary, on='ManagerID', how="left")

result = result[result['Salary'] > result['ManagerSalary']]

print(result[["Employee","Salary","ManagerSalary"]])
```

```python
#Calculate Department Salary Percentage Contribution

import pandas as pd

df = pd.DataFrame({
    "Employee":["A","B","C","D","E"],
    "Department":["HR","HR","IT","IT","IT"],
    "Salary":[50000,70000,90000,80000,60000]
})
```

```python
dept_total = df.groupby('Department')['Salary'].transform('sum')

df['Contribution_%'] = (df['Salary']/dept_total * 100).round(2)

print(df)
```

```python
#Find the Monthly Sales Growth Percentage

import pandas as pd

sales = pd.DataFrame({
    "Month":["Jan","Feb","Mar","Apr","May"],
    "Sales":[1000,1200,1500,1300,1800]
})
```

```python
sales['Growth_%'] = sales['Sales'].pct_change() * 100
print(sales)
```

```python
#Find Consecutive Orders Placed by Customers

import pandas as pd

df = pd.DataFrame({
    "Customer":[101,101,101,102,102,103],
    "Date":[
        "2024-01-01",
        "2024-01-02",
        "2024-01-05",
        "2024-01-10",
        "2024-01-11",
        "2024-01-20"
    ]
})

df["Date"] = pd.to_datetime(df["Date"])
```

```python
df = df.sort_values(['Customer','Date'])

df['Difference'] = df.groupby('Customer')['Date'].diff()

result = df[df['Difference'] == pd.Timedelta(days=1)]
print(result)
```

```python
#Create Department-wise Summary Report
import pandas as pd

df = pd.DataFrame({
    "Department":["HR","HR","IT","IT","Sales","Sales"],
    "Salary":[50000,70000,80000,90000,40000,60000],
    "Performance":[80,90,95,88,70,82]
})
```

```python
summary = (
    df.groupby('Department')
    .agg(
        Employees=('Salary','count'),
        Avg_Salary=('Salary','mean'),
        Max_Salary=('Salary','max'),
        Total_Salary=('Salary','sum'),
        Avg_Performance=('Performance','mean')
    )
)

print(summary)
```

```python
#Merge Customer and Order Tables and Calculate Total Spending
import pandas as pd

customers = pd.DataFrame({
    "CustomerID":[1,2,3],
    "Name":["Alice","Bob","Charlie"]
})

orders = pd.DataFrame({
    "CustomerID":[1,1,2,3,3],
    "Amount":[500,700,600,900,100]
})
```

```python
order_summary = (
    orders.groupby("CustomerID")["Amount"]
          .sum()
          .reset_index(name="Total_Spending")
)

result = customers.merge(
    order_summary,
    on="CustomerID",
    how="left"
)

print(result)
```
