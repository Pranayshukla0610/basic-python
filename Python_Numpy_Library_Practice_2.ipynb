```python
#Generate Employee IDs Using arange()

import numpy as np

employee_ids = np.arange(1001,1101)
print(employee_ids)
```

```python
#Generate time slots from 9 AM to 6 PM.

hours = np.arange(9,19)
print(hours)
```

```python
#Generate Learning Rate Values Using linspace() between 0.001 and 0.1

learning_rates = np.linspace(0.001, 0.1, 10)
print(learning_rates)
```

```python
#Generate 20 equally spaced temperatures between 20°C and 80°C

temperature = np.linspace(20,80,20)
print(temperature)
```

```python
#Generate Exponential Growth Values, create power of 10

growth = np.logspace(1, 10, 10)
print(growth)
```

```python
#Initialize a Neural Network Weight Matrix 50 neurons and 20 features

weights = np.zeros((50,20))
print(weights.shape)
```

```python
#Create attendance for 40 students for 5 days

attendance = np.ones((40,5),dtype=int)
print(attendance)
```

```python
#Initialize probabilities to 1 for 8x8

probabilities = np.ones((8,8))
print(probabilities)
```

```python
#Create Salary Matrix, Every employee gets 50000 in 20x12 salary matrix

salary = np.full((20,12),50000)
print(salary)
```

```python
#Every student initially gets 100, create 30 students 5 subjects

marks = np.full((30,5),100)
print(marks)
```

```python
#Identity Matrix for Matrix Multiplication

identity = np.eye(5)
print(identity)
```

```python
#Multiply matrix by identity matrix.

A = np.array([
    [2,3],
    [4,5]
])

I = np.eye(2)

result = A @ I
print(result)
```

```python
#Create Lower Triangular Matrix

matrix = np.tri(6)
print(matrix)
```

```python
#Generate Coordinate Grid

x = np.arange(1,6)
y = np.arange(1,6)

X, Y = np.meshgrid(x,y)

print("X")
print(X)

print("Y")
print(Y)
```

```python
#Generate Distance Matrix

import numpy as np

x = np.arange(-3,4)

y = np.arange(-3,4)

X,Y = np.meshgrid(x,y)

distance = np.sqrt(X**2 + Y**2)

print(np.round(distance,2))
```
