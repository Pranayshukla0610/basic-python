```python
#Find the Largest Difference Between Two Elements

arr = list(map(int, input("Enter an array:").split()))

min_element = arr[0]
max_diff = 0

for i in range(1,len(arr)):
  if arr[i] - min_element > max_diff:
    max_diff = arr[i] - min_element

  if arr[i] < min_element:
    min_element = arr[i]

print("Maximum Difference:",max_diff)
```

```python
#Find Common Elements in Three Arrays

arr1 = list(map(int, input("Enter first array:").split()))
arr2 = list(map(int, input("Enter second array").split()))
arr3 = list(map(int, input("Enter third array").split()))

common = []

for num in arr1:
  if num in arr2 and num in arr3 and num not in common:
    common.append(num)

print(common)
```

```python
#Count Even and Odd Numbers in an Array

arr = list(map(int, input("Enter an array:").split()))

even = 0
odd = 0

for num in arr:
  if num % 2 == 0:
    even += 1
  else:
    odd += 1

print("Even Number Count:",even)
print("Odd Number Count:",odd)
```

```python
#Find the Maximum Product of Two Elements

arr = list(map(int, input("Enter an array:").split()))

maximum = arr[0] * arr[1]

for i in range(len(arr)):
  for j in range(i+1,len(arr)):
    if arr[i] * arr[j] > maximum:
      maximum = arr[i] * arr[j]

print("Maximum Product:",maximum)
```

```python
#Check if an Array Contains Only Unique Elements

arr = list(map(int, input("Enter an array:").split()))

if len(arr) == len(set(arr)):
  print("All Elements are Unique")
else:
  print("Array Contain Duplicate Elements")
```
