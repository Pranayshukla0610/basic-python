```python
#Find the Top 2 Highest Values in Every Row

import numpy as np

arr = np.array([
    [12, 45, 67, 23],
    [89, 34, 76, 90],
    [11, 56, 78, 44]
])
```

```python
sorted_arr = np.sort(arr, axis=1)

top2 = sorted_arr[:,-2:][:,::-1]

print("Top 2 values of each row")
print(top2)
```

```python
#Replace Maximum Value of Every Row with Zero

arr = np.array([
    [5, 8, 2],
    [9, 1, 7],
    [4, 6, 3]
])
```

```python
row_max_index = np.argmax(arr, axis=1)

arr[np.arange(arr.shape[0]), row_max_index] = 0

print(arr)
```

```python
import numpy as np

arr = np.array([
    [10, 200],
    [20, 300],
    [30, 400],
    [40, 500]
])
```

```python
minimum = arr.min(axis=0)
maximum = arr.max(axis=0)

normalised = (arr - minimum) / (maximum - minimum)

print(normalised)
```

```python
#Find Duplicate Elements and Their Counts

import numpy as np

arr = np.array([2,5,3,2,7,5,9,2,5])
```

```python
unique, counts = np.unique(arr, return_counts=True)

duplicates = unique[counts > 1]
duplicate_counts = counts[counts > 1]

print("Duplicate elements:", duplicates)
print("Their counts:", duplicate_counts)
```

```python
#Reverse Alternate Rows

import numpy as np

arr = np.arange(1,26).reshape(5,5)

print(arr)
```

```python
arr = np.arange(1,26).reshape(5,5)

arr[1::2] = arr[1::2, ::-1]
print(arr)
```

```python
#Find Common and Unique Elements

import numpy as np

arr1 = np.array([1,2,3,4,5])
arr2 = np.array([3,4,5,6,7])

common = np.intersect1d(arr1, arr2)
only_a = np.setdiff1d(arr1, arr2)
only_b = np.setdiff1d(arr2, arr1)

print("Common Element:",common)
print("Unique Element in A:", only_a)
print("Unique Element in B:", only_b)
```

```python
union = np.union1d(arr1, arr2)
print(union)
```

```python
#Sort Matrix Based on Second Column

import numpy as np

arr = np.array([
    [101,90],
    [102,75],
    [103,95],
    [104,88]
])
```

```python
import numpy as np

sales = np.array([
    [101,50000,6000],
    [102,65000,7000],
    [103,40000,5000],
    [104,70000,8000],
    [105,55000,6500]
])

# Columns:
# Employee_ID | Sales | Bonus
```

```python
employee = sales[:,0]
sales_amount = sales[:,1]
bonus = sales[:,2]

print("Total Sales:",sales_amount.sum())
print("Average Sales:",sales_amount.mean())
print("Highest Sales:",sales_amount.max())
print("Lowest Sales:",sales_amount.min())
```

```python
highest_employee = employee[np.argmax(sales_amount)]
print("Highest Sales Employee:",highest_employee)
```

```python
print("Total bonus:",bonus.sum())
bonus_percent = np.round((bonus/sales_amount)*100,2)
print("Bonus Percentage:",bonus_percent)
```

```python
rank = sales_amount.argsort()[::-1]
print("Sales Ranking:")
print(sales[rank])
```

```python
above_average = sales[sales_amount > sales_amount.mean()]
print("Employee Above Average Sales:")
print(above_average)
```

```python
cumulative = np.cumsum(sales_amount)
print("Cumulative Sales:",cumulative)
```

```python
#Calculate Row-wise Percentage Contribution

import numpy as np

sales = np.array([
    [100,200,300],
    [150,250,100],
    [400,300,300]
])
```

```python
row_total = sales.sum(axis=1, keepdims=True)
percentage = (sales / row_total) * 100
print(np.round(percentage,2))
```
