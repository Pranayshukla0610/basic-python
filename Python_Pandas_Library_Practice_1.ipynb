```python
#Top 3 Highest Paid Employees in Each Department

import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Department": ["HR", "HR", "IT", "IT", "IT", "Sales", "Sales", "Sales", "HR", "IT"],
    "Salary": [50000, 65000, 90000, 85000, 75000, 40000, 70000, 68000, 72000, 88000]
})

print(df)
```

```python
df['Rank'] = df.groupby('Department')['Salary'].rank(method='dense',ascending=False)
result = df[df['Rank'] <= 3].sort_values(['Department','Rank'])

print(result)
```

```python
#Department Average Salary and Employees Above Average

import pandas as pd

df = pd.DataFrame({
    "Employee":["A","B","C","D","E","F"],
    "Department":["HR","HR","IT","IT","Sales","Sales"],
    "Salary":[50000,70000,80000,60000,45000,65000]
})

print(df)
```

```python
df['Dept_avg'] = df.groupby('Department')['Salary'].transform('mean')

result = df[df['Salary'] > df['Dept_avg']]

print(result)
```

```python
#Find Duplicate Transactions and Count Them

import pandas as pd

df = pd.DataFrame({
    "Transaction_ID":[101,102,102,103,104,104,104,105],
    "Amount":[500,700,700,600,1000,1000,1000,900]
})

print(df)
```

```python
duplicates = df[df.duplicated(subset='Transaction_ID',keep=False)]

print("Duplicate Transactions:")
print(duplicates)

count = df.groupby("Transaction_ID").size().reset_index(name='Count')

print("\nTransaction Counts:")
print(count)

repeated = count[count["Count"] > 1]

print("\nRepeated Transactions:")
print(repeated)
```

```python
#Department-wise Highest Salary Employee

import pandas as pd

df = pd.DataFrame({
    "Employee":["A","B","C","D","E","F"],
    "Department":["HR","HR","IT","IT","Sales","Sales"],
    "Salary":[50000,70000,90000,80000,45000,65000]
})

print(df)
```

```python
idx = df.groupby('Department')['Salary'].idxmax()

result = df.loc[idx]

print(result)
```

```python
#Merge Customer, Orders, and Payments

import pandas as pd

customers = pd.DataFrame({
    "Customer_ID":[1,2,3,4],
    "Customer_Name":["Alice","Bob","Charlie","David"]
})

orders = pd.DataFrame({
    "Customer_ID":[1,1,2,3],
    "Order_Amount":[500,700,1000,1200]
})

payments = pd.DataFrame({
    "Customer_ID":[1,2,3],
    "Payment":[1000,800,1000]
})
```

```python
order_summary = (orders.groupby("Customer_ID")['Order_Amount'].sum().reset_index(name="Total_Orders"))

report= customers.merge(order_summary, on='Customer_ID',how='left')

report = report.merge(payments, on='Customer_ID', how='left')

report[['Total_Orders','Payment']] = report[['Total_Orders','Payment']].fillna(0)

report['Pending_Amount'] = report['Total_Orders'] - report['Payment']

print(report)
```
