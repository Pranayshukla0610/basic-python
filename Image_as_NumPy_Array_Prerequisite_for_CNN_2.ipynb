```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]
image = Image.open(filename)
print(image)
```

```python
image_array = np.array(image)
```

```python
print(type(image_array))
print(image_array.shape)
```

```python
print("Image shape:", image_array.shape)
```

```python
height = image_array.shape[0]
width = image_array.shape[1]
channels = image_array.shape[2]

print("Height:", height)
print("Width:", width)
print("Channels:", channels)
```

```python
red_channel = image_array[:, :, 0]
green_channel = image_array[:, :, 1]
blue_channel = image_array[:, :, 2]
```

```python
print("Red min:", red_channel.min())
print("Red max:", red_channel.max())

print("Green min:", green_channel.min())
print("Green max:", green_channel.max())

print("Blue min:", blue_channel.min())
print("Blue max:", blue_channel.max())
```

```python
plt.imshow(red_channel, cmap="gray")
plt.title("Red Channel")
plt.axis("off")
plt.show()
```

```python
bright_image = image_array.astype(np.int16) + 50

bright_image = np.clip(bright_image, 0, 255)

bright_image = bright_image.astype(np.uint8)
```

```python
plt.imshow(bright_image)
plt.axis("off")
plt.show()
```

```python
dark_image = image_array.astype(np.int16) - 50

dark_image = np.clip(dark_image, 0, 255)

dark_image = dark_image.astype(np.uint8)
```

```python
plt.imshow(dark_image)
plt.axis("off")
plt.show()
```
