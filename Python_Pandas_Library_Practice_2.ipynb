```python
#Region-wise Product Sales Dashboard using Pivot Table

import pandas as pd

df = pd.DataFrame({
    "Region": ["North","North","South","South","East","East","West","West"],
    "Product": ["Laptop","Mobile","Laptop","Mobile","Laptop","Mobile","Laptop","Mobile"],
    "Sales": [120000,80000,95000,70000,110000,75000,105000,85000],
    "Profit": [18000,12000,15000,10000,16000,11000,17000,13000]
})

print(df)
```

```python
pivot = pd.pivot_table(
    df,
    values=["Sales","Profit"],
    index="Region",
    columns="Product",
    aggfunc={
        "Sales":"sum",
        "Profit":"mean"
    },
    fill_value=0
)

print(pivot)
```

```python
#Convert Wide Data to Long Format using Melt

import pandas as pd

df = pd.DataFrame({
    "Employee":["A","B","C"],
    "January":[50000,60000,70000],
    "February":[52000,61000,71000],
    "March":[53000,62000,73000]
})

print(df)
```

```python
result = pd.melt(
    df,
    id_vars=['Employee'],
    var_name='Month',
    value_name='Salary'
)

print(result)
```

```python
#Create MultiIndex and Slice Data

import pandas as pd

df = pd.DataFrame({
    "Country":["India","India","India","USA","USA"],
    "State":["UP","UP","Delhi","Texas","California"],
    "City":["Kanpur","Lucknow","New Delhi","Dallas","Los Angeles"],
    "Population":[3000000,2800000,19000000,1300000,4000000]
})

print(df)
```

```python
df = df.set_index(['Country','State'])
print(df)
```

```python
#Select India
print(df.loc["India"])
```

```python
#Select UP
print(df.loc['India','UP'])
```

```python
df = df.reset_index()
print(df)
```

```python
#Sales Analysis using Stack and Unstack

import pandas as pd

sales = pd.DataFrame({
    "Laptop":[120,140],
    "Mobile":[90,110],
    "Tablet":[40,55]
}, index=["North","South"])

print(sales)
```

```python
stacked = sales.stack()
print(stacked)
```

```python
unstacked = stacked.unstack()
print(unstacked)
```

```python
#Customer Purchase Analysis using Crosstab and Explode

import pandas as pd

df = pd.DataFrame({
    "Customer":["A","B","C","D"],
    "Products":[
        ["Laptop","Mouse"],
        ["Laptop","Keyboard"],
        ["Mouse","Keyboard"],
        ["Laptop","Mouse","Keyboard"]
    ]
})

print(df)
```

```python
exploded = df.explode("Products")
print(exploded)
```

```python
print(exploded['Products'].value_counts())
```

```python
table = pd.crosstab(
    exploded['Customer'],
    exploded['Products']
)

print(table)
```
