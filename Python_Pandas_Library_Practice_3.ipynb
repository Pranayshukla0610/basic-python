```python
#Find Employees with the 2nd Highest Salary in Each Department

import pandas as pd

df = pd.DataFrame({
    "Employee":["A","B","C","D","E","F","G","H"],
    "Department":["HR","HR","HR","IT","IT","IT","Sales","Sales"],
    "Salary":[50000,70000,65000,90000,85000,80000,45000,60000]
})

print(df)
```

```python
df['Rank'] = df.groupby('Department')['Salary'].rank(ascending=False)

result = df[df['Rank'] == 2]
```

```python
#Find Departments Having Average Salary Greater Than 65000
df['avg_sal'] = df.groupby('Department')['Salary'].transform('mean')
result = df[df['avg_sal'] > 65000]
print(result)
```

```python
#Find Employees Earning More Than Company Average

import pandas as pd

df = pd.DataFrame({
    "Employee":["A","B","C","D","E"],
    "Salary":[50000,70000,80000,60000,90000]
})
```

```python
df['avg_sal'] = df['Salary'].mean()

result = df[df['Salary'] > df['avg_sal']]
print(result)
```

```python
#Find Department with Maximum Total Salary

import pandas as pd

df = pd.DataFrame({
    "Department":["HR","HR","IT","IT","Sales","Sales"],
    "Salary":[50000,70000,90000,80000,45000,55000]
})
```

```python
dept_salary = (
    df.groupby("Department")["Salary"]
      .sum()
)

print(dept_salary)

highest = dept_salary.idxmax()

print(highest)
```

```python
#Find Employees Whose Salary is Equal to Department Maximum

import pandas as pd

df = pd.DataFrame({
    "Employee":["A","B","C","D","E","F"],
    "Department":["HR","HR","IT","IT","Sales","Sales"],
    "Salary":[50000,70000,90000,80000,45000,55000]
})
```

```python
df['max_sal'] = df.groupby('Department')['Salary'].transform('max')

result = df[df['Salary'] == df['max_sal']]

print(result)
```

```python
#Find Total Orders and Average Order Value per Customer

import pandas as pd

orders = pd.DataFrame({
    "Customer":["A","A","B","B","B","C"],
    "OrderValue":[200,300,500,600,400,700]
})
```

```python
summary = (
    orders.groupby('Customer').
    agg(
        Total_Orders = ('OrderValue','count'),
        Average_Order_Value = ('OrderValue','mean')
    )
)

print(summary)
```

```python
#Find Customers Having More Than Two Orders

import pandas as pd

orders = pd.DataFrame({
    "Customer":["A","A","B","B","B","C","D","D","D","D"],
    "Amount":[100,200,300,400,500,600,700,800,900,1000]
})
```

```python
result = orders.groupby('Customer').filter(lambda x:len(x) > 2)
print(result)
```

```python
#Merge Two DataFrames and Find Customers Without Orders

import pandas as pd

customers = pd.DataFrame({
    "CustomerID":[1,2,3,4,5]
})

orders = pd.DataFrame({
    "CustomerID":[1,2,4]
})
```

```python
result = customers.merge(
    orders,
    on = 'CustomerID',
    how='left',
    indicator=True
)
result = result[result['_merge']=='left_only']
print(result)
```

```python
#Find Percentage Contribution of Each Employee to Department Salary

import pandas as pd

df = pd.DataFrame({
    "Employee":["A","B","C","D","E"],
    "Department":["HR","HR","IT","IT","IT"],
    "Salary":[50000,70000,90000,80000,60000]
})
```

```python
dept_total = df.groupby('Department')['Salary'].transform('sum')

df['Contribution_%'] = (df['Salary'] / dept_total) * 100

print(df)
```
