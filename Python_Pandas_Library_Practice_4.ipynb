```python
#Find Employees Whose Salary is More Than 20% Above Department Average

import pandas as pd

df = pd.DataFrame({
    "Employee":["A","B","C","D","E","F"],
    "Department":["HR","HR","IT","IT","Sales","Sales"],
    "Salary":[50000,70000,80000,90000,45000,60000]
})
```

```python
df["dept_avg"] = df.groupby('Department')['Salary'].transform("mean")
result = df[df['Salary'] > 0.2 *df['dept_avg']]
print(result)
```

```python
#Find First Employee Hired in Every Department

import pandas as pd

df = pd.DataFrame({
    "Employee":["A","B","C","D","E","F"],
    "Department":["HR","HR","IT","IT","Sales","Sales"],
    "Joining_Date":[
        "2022-03-01",
        "2021-05-10",
        "2020-01-15",
        "2023-04-20",
        "2019-07-01",
        "2022-08-15"
    ]
})

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
df
```

```python
idx = df.groupby('Department')['Joining_Date'].idxmin()

print(df.loc[idx])
```

```python
#Calculate Running Total of Salary Inside Each Department

import pandas as pd

df = pd.DataFrame({
    "Department":["HR","HR","HR","IT","IT","Sales"],
    "Employee":["A","B","C","D","E","F"],
    "Salary":[40000,50000,60000,70000,80000,90000]
})
df
```

```python
df['running_tot'] = df.groupby('Department')['Salary'].cumsum()
print(df)
```

```python
#Find Employee Number Inside Each Department

import pandas as pd

df = pd.DataFrame({
    "Department":["HR","HR","HR","IT","IT","Sales"],
    "Employee":["A","B","C","D","E","F"]
})
```

```python
df['emp_num'] = df.groupby('Department')['Employee'].cumcount()+1
print(df)
```

```python
#Number Every Department

import pandas as pd

df = pd.DataFrame({
    "Department":["HR","HR","IT","IT","Sales","Finance"]
})
```

```python
df["Dept_Id"] = df.groupby('Department').ngroup()
print(df)
```

```python
#Replace Department Names Using Mapping

import pandas as pd

df = pd.DataFrame({
    "Department":["HR","IT","Sales","Finance"]
})
df
```

```python
mapping = {
    "HR":"Human Resources",
    "IT":"Information Technology",
    "Sales":"Sales Team",
    "Finance":"Finance Team"
}

df['Department'] = df['Department'].map(mapping)
print(df)
```

```python
#Find Department Having Maximum Employees

import pandas as pd

df = pd.DataFrame({
    "Department":[
        "HR","HR","HR",
        "IT","IT",
        "Sales","Sales","Sales","Sales"
    ]
})
```

```python
count = df["Department"].value_counts()

print(count)

print("Department:", count.idxmax())
```

```python
#Find Departments Having At Least Three Employees

import pandas as pd

df = pd.DataFrame({
    "Department":[
        "HR","HR","HR",
        "IT","IT",
        "Sales","Sales","Sales",
        "Finance"
    ]
})
```

```python
results = df.groupby('Department').filter(lambda x: len(x) >= 3)
print(results)
```

```python
#Merge Employee and Department Tables

import pandas as pd

employees = pd.DataFrame({
    "EmpID":[101,102,103],
    "DepartmentID":[1,2,1]
})

departments = pd.DataFrame({
    "DepartmentID":[1,2],
    "Department":["HR","IT"]
})
```

```python
result = employees.merge(
    departments,
    on='DepartmentID',
    how='inner'
)
print(result)
```
