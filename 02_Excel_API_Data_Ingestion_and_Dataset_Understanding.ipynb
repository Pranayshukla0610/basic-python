API Error Handling

```python
try:

    response = requests.get(
        url,
        timeout=10
    )

    response.raise_for_status()

    data = response.json()

    api_df = pd.DataFrame(data)

    print(
        "Data successfully ingested"
    )

except requests.exceptions.RequestException as e:

    print(
        "API request failed:",
        e
    )
```

API Parameters

```python
params = {
    "page": 1,
    "limit": 100
}
```

```python
response = requests.get(
    url,
    params=params,
    timeout=10
)
```

API Headers

```python
headers = {
    "Authorization": "Bearer YOUR_API_KEY",
    "Accept": "application/json"
}
```

```python
response = requests.get(
    url,
    headers=headers,
    timeout=10
)
```

API Pagination

```python
all_records = []

for page in range(1, 6):

    params = {
        "page": page,
        "limit": 100
    }

    response = requests.get(
        url,
        params=params,
        timeout=10
    )

    response.raise_for_status()

    data = response.json()

    all_records.extend(data)
```

```python
api_df = pd.DataFrame(
    all_records
)
```
